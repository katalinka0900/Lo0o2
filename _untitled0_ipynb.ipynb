Мы хотим определить, влияют ли следующие показатели на численность студентов среднего профессионального образования (СПО):
1. Средняя начисленная заработная плата (`Wages`).
2. Численность трудоспособного населения (`WorkAge`).
3. Удельный вес занятого населения со средним профессиональным образованием (`SecEdu`).

Также нужно выяснить, можно ли использовать показатель "численность населения" (`Popul`).

Для анализа зависимости между переменными я воспользуюсь **линейной регрессией**. Этот метод позволяет:
- Оценить влияние каждого предиктора на целевую переменную.
- Проверить статистическую значимость предикторов с помощью p-values.
- Оценить качество модели с помощью коэффициента детерминации \( R^2 \).

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Загрузка данных
data = pd.read_excel('Данные для задания.xlsx', sheet_name='Данные')
```

```python
# Подготовка данных
X = data[['Wages', 'WorkAge', 'SecEdu']]  # Предикторы
y = data['SecStudent']  # Целевая переменная

# Добавим константу для учета свободного коэффициента
X = sm.add_constant(X)
```

```python
# Построение модели
model = sm.OLS(y, X).fit()

# Вывод результатов
print(model.summary())
```



 **Выводы**

### 1. Влияние предикторов на численность студентов СПО
1. Средняя начисленная заработная плата (`Wages`):

Отрицательный коэффициент (-0.0004) указывает на то,
что с ростом средней зарплаты численность студентов СПО немного уменьшается.

   Это может быть связано с тем, что в регионах с высокой зарплатой люди чаще выбирают высшее образование или другие формы обучения, а не СПО.

2. Численность трудоспособного населения (`WorkAge`):

Отрицательный коэффициент (-0.0025) не является статистически значимым (p-value = 0.437).

Это означает, что численность трудоспособного населения не оказывает значимого влияния на численность студентов СПО.

3. Удельный вес занятого населения со СПО (`SecEdu`):

Положительный коэффициент (0.4888) не является статистически значимым (p-value = 0.382).

 Это означает, что удельный вес занятого населения со СПО не оказывает значимого влияния на численность студентов СПО.








```python
# Оценка модели
print("R^2:", model.rsquared)
print("Adjusted R^2:", model.rsquared_adj)
print("P-values:", model.pvalues)
```

```python
# Проверка использования показателя "численность населения"
X_with_popul = data[['Wages', 'WorkAge', 'SecEdu', 'Popul']]
X_with_popul = sm.add_constant(X_with_popul)
model_with_popul = sm.OLS(y, X_with_popul).fit()
print(model_with_popul.summary())
```
